# Tipus de dades
Hi ha diferents estructures de dades a Python com els diccionaris, sets, sequence i iteradors. A continuació, coneixerem més detalladament cadascun d'ells.

## Diccionaris

Els diccionaris són estructures on es poden emmagatzemar diferents tipus de dades (`string`, `integer`, `float`, `boolean`, `llistes`) en una mateixa estructura.

La característica fonamental dels diccionaris és que l'emmagatzematge de les dades es realitza per parells, és a dir, donada una clau, obtenim un valor. Per definir un diccionari, utilitzem `:` per separar la clau del valor. A més, s'ocupen claus ```{``` ```}``` per definir l'inici i el final de la definició del diccionari.

 ```diccionari = {key:value}```

A l'exemple següent emmagatzemarem les dades d'una persona considerant incloure els diferents tipus de variables.

````python
dades = {
 "nom" : "Mario",
 "edat" : 33,
 "talla" : 1.85,
 "colors preferits" : ["negre","blau"],
 "treballa?" : True
}
print(dades)
````

**Nota:** Python no permet diversos valors per a una mateixa clau. En el cas que ocorri això, es tindrà en compte l'últim valor que s'hagi assignat a aquesta clau.

````python
dades = {
 "nom" : "Mario",
 "nom" : "Carla",
 "edat" : 33,
 "nom" : "Xavier",
 "talla" : 1.85
}
print(dades)
````

### Mètodes i propietats d'un diccionari

#### Visualització

Per visualitzar tota la llista podem utilitzar ```.item()``` .

````python
dades.items()
````

Per visualitzar totes les claus que conté el diccionari podem utilitzar ```.keys()``` .

````python
dades.keys()
````

Per visualitzar tots els valors que conté el diccionari podem utilitzar ```.values()``` .

````python
dades.values()
````

#### Longitud d'un diccionari


Per conèixer la longitud d'un diccionari utilitzarem el mètode ```len()```.

Utilitzarem el diccionari ```dades```.

````python
print(len(dades))
````

#### Comprovar si existeix certa clau al diccionari

Sabem que les claus que conformen el nostre diccionari `dades` són:
```['nom', 'edat', 'talla']```, i obtindrem respostes boleeanes ```True o False```.

````python
print("nom" in dades)
````

````python
print("Hola" in dades)
````

#### Afegir elements a un diccionari

Per afegir un element a un diccionari s'ha d'assignar el valor de l'element al nom del diccionari amb el nom de la clau entre claudàtors. Per exemple:

````python
dades["treballa?"] = True
print(dades)
````

#### Reassignar un valor en un diccionari


Utilitzarem el mateix diccionari ```dades```, i assignarem un altre valor en ```edat```.

````python
dades["edat"] = 28
print(dades)
````

També podem utilitzar el mètode ```update()```.

````python
dades.update({"edat" : 30})
print(dades)
````

#### Eliminar elements d'un diccionari


Per eliminar es pot utilitzar el mètode ```.pop()```.

Eliminar un element mitjançant la seva clau.

````python
dades.pop("treballa?")
print(dades)
````

Eliminar l'últim element de la llista de claus del diccionari.



````python
dades.popitem()
print(dades)
````

També podem utilitzar el mètode ```.del()``` per eliminar una clau amb el seu respectiu valor.

````python
del dades['nom']
print(dades)
````

### Sets


Els ```Set``` són una estructura de dades que permet emmagatzemar un conjunt de valors en una sola variable. Per definir el conjunt de valors d'un `Set` utilitzem les claus ```{}```.

Els `Set` no es poden editar, no mantenen l'ordre dels elements i, si es troba una dada repetida, serà considerat una sola vegada, com es mostra a continuació a l'exemple següent.


````python
el_nostre_set = {"Això és", "Això és", 1, "set", "podem guardar diferents dades", 3.1416}
print(el_nostre_set)
````

### Mètodes i propietats d'un `Set`

#### Visualització

Per poder visualitzar element per element que conté el set podem realitzar una impressió amb un bucle ```for```.

````python
for i in el_nostre_set:
  print(i)
````

#### Longitud d'un `Set`

Per conèixer la longitud del nostre `Set` utilitzem el mètode ```len()```.

````python
len(el_nostre_set)
````

#### Comprovar si un valor es troba al nostre `Set`

En el cas dels `Set`, simplement s'introdueix l'element que desitgem conèixer si pertany o no al set i obtindrem respostes booleanes.

````python
"Això és" in el_nostre_set
````

````python
"Això" in el_nostre_set
````

#### Buidar el `Set`

Per buidar el set utilitzem el mètode ```.clear()```.

````python
el_nostre_set.clear()
print(el_nostre_set)
````

#### Afegir elements al nostre `Set`

Per afegir un element utilitzem el mètode ```.add()```.

````python
el_nostre_set.add("Hola")
print(el_nostre_set)
````

Per afegir un nou `Set` ja creat a un altre `Set`, utilitzem el mètode ```.update()```.

````python
nou_set = {"Blau", "Groc", "Vermell"}
#Primer esmentem el set al qual agregarem el nou set.
el_nostre_set.update(nou_set)
print(el_nostre_set)
````

#### Combinar `Set`

Podem combinar `Set` amb el mètode ```.union()```, és necessari que els `Set` combinats es guardin en un nou `Set`.

````python
nou_set = {"Maduixa", "Cirera", "Banana"}
set_final = el_nostre_set.unio(nou_set)
print(set_final)
````

#### Eliminar elements del nostre `Set`


Per eliminar algun element del `Set` podem utilitzar mètodes com ```.discard()``` i ```.remove()```. Aquests funcionen especificant l'element que desitgem eliminar, i si l'element no existeix dins del `Set` ens generarà un error; també podem utilitzar el mètode ```.pop()```, que es diferencia dels anteriors degut a que aquest mètode elimina de manera **aleatòria** els elements dins del `Set`.

````python
set_final.discard("Groc")
print(set_final)
````

````python
set_final.remove("Blau")
print(set_final)
````

````python
set_final.pop()
print(set_final)
````

Si es desitja obtenir informació addicional es pot accedir mitjançant l'ordre ```help(set)```.

````python
help(set)
````

### Sequences

Els ```sequence``` són conjunts de dades que, a diferència dels ```set```, mantenen un ordre i al seu torn admeten elements repetits. La definició dels `sequence` es realitza utilitzant claudàtors ```[``` ```]```.

Es classifiquen en mutables, com `list` i `bytearray`; i immutables, com `string`, `tuple`, `range` i `byte`.

D'igual manera poden ser de tipus homogeni o heterogeni, és a dir, poden emmagatzemar els mateixos tipus de dades, o diferents tipus.

A continuació, tenim un ```sequence``` homogeni.

````python
my_sequence = [1, 2, 3 , 4 , 5] #Totes les dades són nombres enters.
print(my_sequence)
````

A continuació, tenim un ```sequence``` heterogeni.

````python
mysequence = ["Hola", "això és un sequence", 1.8, 10,"Hola"]
print(mysequence)
````

### Mètodes i propietats d'un sequence

#### Visualització

````python
mysequence[0]
````

````python
mysequence[3]
````

#### Longitud d'un sequence

````python
print(len(mysequence))
````

#### Comprovar si un valor es troba al nostre sequence

````python
"Hola" in mysequence
````

#### Buidar sequence

````python
mysequence.clear()
print(mysequence)
````

#### Afegir elements al nostre sequence

Per afegir elements al nostre ```sequence``` utilitzem el mètode ```.append()```.

````python
mysequence.append("Sequence")
print(mysequence)
````

````python
mysequence.append(1.85)
print(mysequence)
````

#### Combinar sequence

Per combinar elements al nostre ```sequence``` utilitzem el mètode ```.extend()```.

````python
altresequence = [1, 5, 10, 20, "A"]
mysequence.extend(altresequence)
print(mysequence)
````

#### Invertir el nostre sequence

````python
mysequence.reverse()
print(mysequence)
````

#### Eliminar elements del nostre sequence

Per eliminar algun element del nostre ```sequence``` podem utilitzar mètodes com ```.remove()``` i ```.pop()```. Utilitzem ```.remove()``` per eliminar l'element que desitgem esmentant-lo i ```.pop()``` ens permet eliminar l'element d'acord a l'índex donat o eliminant l'últim element si no s'indica l'índex.

````python
mysequence.remove("Sequence")
print(mysequence)
````

````python
mysequence.pop(0)
print(mysequence)
````

````python
mysequence.pop()
print(mysequence)
````

### Iterators

Els ```iteradors``` són els elements que obtenim dels objectes iterables com són les llistes, tuples, diccionaris i cadenes de text on es pot iterar, és a dir, es poden recórrer de valor en valor.

Per obtenir un iterador utilitzem ```iter()``` i ```next()```.

#### Iterador en una llista

````python
my_list = ["Hola", "Hello", 1, 2, True, False, 1.85, 0.75]
print(my_list)
````

````python
my_iterator = iter(my_list)
print(next(my_iterator)) #Primer element
print(next(my_iterator)) #Segon element
print(next(my_iterator)) #Tercer element
print(next(my_iterator)) #Quart element
````

````python
#S'imprimeix des de l'element que ja va ser iterat
for item in my_iterator:
  print(item)
````

#### Iterador en una tupla

````python
my_tuple = ("Groc", "Vermell", "Blau","Verd", 3, 4)
print(my_tuple)
````

````python
my_iterator = iter(my_tuple)
print(next(my_iterator)) #Primer element
print(next(my_iterator)) #Segon element
print(next(my_iterator)) #Tercer element
````

````python
#S'imprimeix des de l'element ja iterat
for item in my_iterator:
  print(item)
````

#### Iterador en un diccionari

````python
my_dic = {
 "nom" : "Mario",
 "edat" : 33,
 "talla" : 1.85,
 "colors preferits" : ["negre","blau"],
 "treballa?" : True
}
print(my_dic)
````

**Nota:** En el cas d'iterar amb els diccionaris només obtindrem les claus que es tinguin en aquest.

````python
my_iterator = iter(my_dic)
print(next(my_iterator)) #Primer element
print(next(my_iterator)) #Segon element
print(next(my_iterator)) #Tercer element
````

````python
#S'imprimeix des de l'element ja iterat
for item in my_iterator:
  print(item)
````

#### Iterador en una cadena de text

````python
my_string = "Hello World"
print(my_string)
````

#### Resum de tipus de dades avançades


***Diccionaris:*** emmagatzema parells de dades mitjançant una clau i un valor; no permet dades repetides.


***Set:*** emmagatzema dades sense mantenir cap ordre, per la qual cosa no manté cap índex dins del set i no permet elements repetits.


***Sequence:*** emmagatzema dades mantenint l'ordre i permet elements repetits.


***Iterator:*** són els elements obtinguts d'objectes iterables com les llistes, diccionaris, tuples, cadenes de text.
