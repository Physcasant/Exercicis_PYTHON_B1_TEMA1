## Extensió de codi mitjançant orientació a objectes

La programació d'orientació a objectes és un paradigma de programació que es basa en el concepte d'objectes, tal com hem vist al tema 4. S'utilitza sovint per estendre el codi d'una manera que és més fàcil d'entendre, mantenir i ampliar.

Els avantatges d'estendre el codi mitjançant aquest paradigma són:

- Permet crear una jerarquia de classes i compartir atributs i mètodes entre elles a través de l'herència.
- Permet crear objectes amb valors únics per als seus atributs, la qual cosa facilita la creació de conjunts d'objectes que comparteixen certes característiques, però també tenen diferències.
- Facilita la reutilització de codi en permetre que els mètodes i atributs d'una classe s'utilitzin en diversos objectes.
- Ajuda a organitzar i dividir el codi en components més manejables i fàcils d'entendre.
- Millora la llegibilitat i el manteniment del codi en permetre que cada classe tingui una responsabilitat clara i definida.

Tot això s'aconsegueix mitjançant l'ús d'herència, sobreescriptura, encapsulament, etc., que es va veure al tema 4.

Hi ha conceptes teòrics addicionals que s'han d'entendre per poder fer una correcta extensió de codi mitjançant l'ús de la programació orientada a objecte, que són: cohesió, abstracció, encapsulament i acoblament.

### Cohesió

La cohesió es refereix a la mesura en què una classe o un mòdul té una única responsabilitat i està altament relacionat amb aquesta responsabilitat. Una classe o mòdul amb alta cohesió és fàcil d'entendre i mantenir, ja que el seu codi està clarament enfocat en un propòsit específic. En dissenyar classes i mòduls amb alta cohesió, es pot millorar la reutilització del codi i la facilitat de manteniment de la teva aplicació.

A Python, pots augmentar la cohesió de les teves classes i mòduls seguint aquests consells:

- Mantingues les classes i mòduls petits i enfocats.
- Assigna responsabilitats clares a cada classe i mòdul.
- Utilitza l'herència i el polimorfisme de manera apropiada.

A continuació, es mostra un exemple de codi.

````python
# La classe Cercle té una única responsabilitat: representar i manipular cercles
class Cercle:
    def __init__(self, radi):
        self.radi = radi

    # El mètode get_area té una responsabilitat clara i definida: calcular i retornar l'àrea d'un cercle
    def get_area(self):
        return 3.14 * self.radi * self.radi

    # El mètode get_perimeter té una responsabilitat clara i definida: calcular i retornar el perímetre d'un cercle
    # Això mostra com el mètode get_perimeter té alta cohesió, ja que està altament relacionat amb el seu propòsit
    def get_perimeter(self):
        return 2 * 3.14 * self.radi

# La classe Cercle manté alta cohesió en tenir una única responsabilitat (representar i manipular cercles)
cercle1 = Cercle(5)
print(cercle1.get_area())
print(cercle1.get_perimeter())
````

En aquest exemple, hem creat una classe `Circle` amb dos mètodes: `get_area` i `get_perimeter`. Cada mètode té una responsabilitat clara i definida: el mètode `get_area` calcula i retorna l'àrea d'un cercle, mentre que el mètode `get_perimeter` calcula i retorna el perímetre d'un cercle. Això mostra com la cohesió pot ajudar a fer que la classe `Circle` sigui fàcil d'entendre i mantenir, ja que el seu codi està clarament enfocat en el càlcul d'àrees i perímetres de cercles.

### Abstracció

L'abstracció és un principi de la programació orientada a objectes que es refereix a l'habilitat de representar només els aspectes rellevants d'un problema o situació sense incloure detalls innecessaris. En utilitzar abstracció pots crear classes que representin conceptes abstractes i després crear objectes concrets a partir d'aquestes classes. Això permet simplificar i resumir informació complexa i millorar la llegibilitat del codi.

A Python, es pot utilitzar l'abstracció de diverses maneres:

- Creant classes abstractes, ja que les classes abstractes s'utilitzen com a plantilles per crear classes concretes que implementin els mètodes abstractes de manera específica. Aquest concepte es va veure al tema 4c (Objectes) i serà fonamental per als pròxims usos d'abstracció.
- Utilitzant interfícies.
- Utilitzant herència.
- Utilitzant polimorfisme.

A continuació, es mostra un exemple d'extensió de codi, mitjançant la implementació d'interfícies mitjançant classes abstractes i herència.

````python
from abc import ABC, abstractmethod

class Cotxe(ABC):   # Interfície mitjançant classe abstracta
    @abstractmethod
    def accelerar(self):
        pass

class Esportiu(Cotxe):    # Implementa de Cotxe mitjançant Herència
    def accelerar(self):
        horse_power = 200
        print(f"El cotxe accelera amb {horse_power} cavalls de força")

class Familiar(Cotxe):    # Implementa de Cotxe mitjançant Herència
    def accelerar(self):
        horse_power = 100
        print(f"El cotxe accelera amb {horse_power} cavalls de força")

class Electric(Cotxe):    # Implementa de Cotxe mitjançant Herència
    def accelerar(self, nivell_bateria):    # Ús de polimorfisme en el mètode accelerar
        bateria = int(nivell_bateria)      # Rep un nombre enter o un text
        print(f"El cotxe accelera amb {bateria}% de bateria restant")

esportiu = Esportiu()    # Instància la classe Esportiu
esportiu.accelerar()

familiar = Familiar()      # Instància la classe Familiar
familiar.accelerar()

elèctric = Electric()    # Instància la classe Electric
elèctric.accelerar(80)     # Anomenat de mètode accelerar amb paràmetre sencer
elèctric.accelerar("95")   # Anomenat de mètode accelerar amb paràmetre tipus text
````

Al codi anterior, es va crear la classe abstracta `Cotxe` i es va heretar a altres classes com `Esportiu`, `Familiar` i `Elèctric`; cada classe té una manera diferent d'implementar el mètode accelerar. A la classe `Elèctric` el mètode accelerar s'implementa de manera diferent, esmentant la bateria en lloc dels cavalls de força, a més, utilitza polimorfisme per funcionar amb valors de paràmetres sencers o tipus text.


### Encapsulació

L'encapsulament és un principi de la programació orientada a objectes que fa referència a la idea d'embolcallar dades i comportaments relacionats en una única unitat anomenada objecte. L'encapsulament permet ocultar els detalls d'implementació d'un objecte i proporcionar una interfície pública per accedir i manipular les dades de l'objecte. D'aquesta manera es concentra en el "què" i no en el "com".

A Python, podem implementar l'encapsulament de dues maneres:

- Utilitzant noms de variables privades: a Python, no existeix una manera de declarar variables privades de manera explícita, però es poden usar noms de variables que comencen amb un guió baix per indicar que són privades i no han de ser accedides directament des de l'exterior. Per exemple: `self._private_var`

- Usant mètodes d'accés: podem crear mètodes públics per accedir i manipular les dades privades d'un objecte. Aquests mètodes es coneixen com a mètodes d'accés i solen tenir noms com get_variable i set_variable. D'aquesta manera, es pot controlar com s'accedeix i modifica la informació privada de l'objecte.

A continuació, es mostra un exemple de com es pot usar l'encapsulament per ocultar els detalls d'implementació d'un objecte:




````python
class BankAccount:
    def __init__(self, balanç=0):
        # Atribut privat _balanç
        self._balanç = balanç

    # Mètode que retorna el valor de _balanç
    def get_balanç(self):
        return self._balanç

    # Mètode que modifica el valor de _balanç
    def set_balanç(self, balanç):
        if balanç >= 0:
            self._balanç = balanç
        else:
            print(f"El balanç no pot ser negatiu, valor: {balanç}")

account = BankAccount(100)
print(f"El balanç és: {account.get_balanç()}")

account.set_balanç(50)
print(f"El nou balanç és: {account.get_balanç()}")

account.set_balanç(-10)
````

A l'exemple es crea una instància de la classe `BankAccount` i s'utilitzen els mètodes `get_balanç` i `set_balanç` per accedir i modificar el saldo del compte representat per l'atribut `_balanç`. Al mètode `set_balanç` s'usa un verificador de valors positius per evitar incongruències. D'aquesta manera es manté un baix acoblament (que veurem a continuació) en la classe, ja que no depèn de classes externes per verificar-lo.

### Acoblament

L'acoblament és una mesura de com de fortament dos mòduls (classes, mètodes, funcions, etc.) estan vinculats entre si. Un alt acoblament significa que els mòduls estan fortament vinculats i depenen l'un de l'altre, mentre que un baix acoblament significa que els mòduls estan poc vinculats i poden funcionar de manera independent.

Un baix acoblament és desitjable en la programació, ja que significa que els mòduls són més fàcils de reutilitzar i mantenir. Per exemple, si tenim una classe amb un alt acoblament amb altres classes, és probable que hàgim de fer canvis en diverses classes si volem modificar alguna part del seu funcionament. En canvi, si tenim una classe amb un baix acoblament, podrem modificar el seu funcionament sense haver de fer canvis en altres classes.

**Com reduir l'acoblament:**

Hi ha diverses maneres de reduir l'acoblament en un codi:
- Utilitzar l'encapsulació. (Vist a l'exemple d'encapsulació).
- Utilitzar abstracció.
- Utilitzar inversió de dependències.

A continuació, es mostra un exemple de codi que té baix acoblament perquè utilitza la inversió de dependències i abstracció. En aquest exemple, es vol crear la classe `Cotxe`, un cotxe pot tenir diferents parts com: motor, suspensió, sistema elèctric, etc. La classe `Cotxe` no ha de dependre directament d'altres classes, sinó que rep una instància d'una interfície que és implementada per altres classes, com la classe `Motor`. D'aquesta manera, la classe `Cotxe` només ha de conèixer la interfície i no ha de conèixer la implementació específica d'altres classes. Si volem implementar la classe `Motor`, sabem que pot ser un motor a gasolina, elèctric, d'hidrogen, etc. Però la classe `Cotxe` no ha de dependre de la classe `Motor`, és a dir, si la implementació de motor elèctric canvia o es descobreix una altra tecnologia per a un motor, no ha d'alterar el funcionament de la classe `Cotxe`, per a això s'usa inversió de dependències.

````python
# Creem una interfície per als motors dels cotxes
class Motor(ABC):
    @abstractmethod
    def engegar(self):
        pass

# Creem una classe concreta per a un motor de gasolina
class MotorGasolina(Motor):
    def engegar(self):
        print("Engegant cotxe amb motor de gasolina")

# Creem una classe concreta per a un motor elèctric
class MotorElectric(Motor):
    def engegar(self):
        print("Engegant cotxe amb motor elèctric")

class MotorHidrogen(Motor):
    def engegar(self):
        print("Engegant cotxe amb motor d'hidrogen")

# Creem la classe Cotxe, que rep una instància de Motor en el seu constructor
# La classe Cotxe no ha de conèixer la implementació específica de la classe Motor, sinó que només ha de conèixer
# la interfície Motor
class Cotxe:
    def __init__(self, motor: Motor):
        self.motor = motor

    def engegar(self):
        self.motor.engegar()

# Creem un objecte de la classe MotorGasolina i el passem com a argument al constructor de la classe Cotxe
cotxe_motor_gasolina = Cotxe(MotorGasolina())
cotxe_motor_gasolina.engegar()

# Creem un objecte de la classe MotorElectric i el passem com a argument al constructor de la classe Cotxe
cotxe_motor_electric = Cotxe(MotorElectric())
cotxe_motor_electric.engegar()
````

## Extensió de codi mitjançant sistema de mòduls i paquets

A Python, podem estendre el nostre codi mitjançant l'ús de mòduls i paquets. Un mòdul és un arxiu que conté codi Python, mentre que un paquet és un conjunt de mòduls organitzats d'alguna manera.

Els mòduls ens permeten dividir el nostre codi en diferents arxius i organitzar-lo de manera més eficient, la qual cosa ens ajuda a mantenir el nostre codi net i ben estructurat. A més, els mòduls ens permeten reutilitzar codi en diferents parts del nostre projecte sense haver de copiar i enganxar el mateix codi en diversos llocs.

Els paquets ens permeten organitzar els nostres mòduls en una estructura de carpetes, la qual cosa ens permet dividir el nostre codi en mòduls més petits i fàcils de mantenir. A més, els paquets ens permeten crear una jerarquia de mòduls, la qual cosa ens ajuda a mantenir una estructura clara i organitzada en projectes més grans.
