# Gestió d'errors


Python permet trobar i solucionar els errors de codi depenent de la seva naturalesa; això pot ser útil per evitar que el programa es bloquegi o acabi inesperadament, i d'aquesta manera es pugui proporcionar una resposta adequada a l'usuari o al sistema.

Per manejar els errors a Python podem fer-ho mitjançant: la lectura dels _TraceBack_, usant sentències `print`, emprant el debugger de l'IDE que estiguem utilitzant i realitzant un correcte maneig d'excepcions amb les ordres `try`, `except`, `else` i  `finally`.

A continuació es presenten les principals tècniques de gestió d'errors.

## Tractament d'errors a Python
### Bàsic = ús de banderes (o _flags_)

El valor de la bandera ha d'inicialitzar-se abans de començar el bucle i ha de canviar el seu estat (valor) dins del cos del bucle.

Aquestes banderes són de gran ajuda per tractar errors a Python, pel fet que ens facilita saber si s'està complint una condició o un procés que té un error, o no funciona de la manera desitjada.

A continuació, es mostra una bandera que permet validar si la variable pren el nou valor.



````python
"""
Problema: es vol aplicar descompte als dies que comencin amb "S".
En aquest cas s'utilitza la bandera per identificar si el descompte va ser o no va ser aplicat.
"""

bandera = False
valor_descompte = 0
condicio = "S"

for dia in ["Dilluns", "Dimarts", "Dimecres", "Dijous", "Divendres", "Dissabte", "Diumenge"]:
  if dia.startswith(condicio):
    valor_descompte = 20
    bandera = True

print("El descompte va ser aplicat?", bandera)
````

### Imprimir per consola
Una altra manera de detectar errors és imprimir per consola utilitzant la funció `print()`. Per exemple, com es mostra a continuació, es pot verificar el valor monitorat en cada iteració d'un bucle for mitjançant un `print()`, i d'aquesta manera detectar l'error.


````python
numero = 4
factorial = 1
for i in range (numero):
    print("S'està multiplicant el nombre ",i," per ", factorial)
    factorial = factorial * i
    print(" i és igual a: ",factorial)
    print("="*15)
print ("El factorial del nombre",numero,"és:",factorial)
````

A l'exemple anterior es pot verificar que l'error està en la primera iteració, ja que es multiplica per zero.

A continuació es té el mateix exemple corregit:

````python
numero = 4
factorial = 1
for i in range (1,numero + 1):
    print("S'està multiplicant el nombre ",i," per ", factorial)
    factorial = factorial * i
    print(" i és igual a: ",factorial)
print ("El factorial del nombre",numero,"és:",factorial)
````

## Excepcions

Les excepcions són errors que ocorren durant l'execució d'un programa, encara que la sintaxi del codi sigui correcta. A Python, aquests errors poden ser manejats mitjançant l'ús de declaracions especials com `try`, `except`, `finally`, etc. No obstant això, sovint ocorren excepcions que no són manejades pel codi i que produeixen missatges d'error, com divisions per zero, ús de variables no declarades, problemes en concatenar cadenes de text, etc.

Exemples:

````python
7 * (1/0)
````

````python
3 + hola * 2
````

````python
'1' + 1
````

Per poder llegir errors com les cel·les mostrades anteriorment, s'ha de començar des de l'última línia.

En aquesta trobarem el nom de l'excepció estàndard, que són identificadors incorporats a l'intèrpret de Python (no són paraules clau reservades). Alguns noms d'excepció es mostren a continuació: ZeroDivisionError, NameError i TypeError.

Seguidament, anem llegint les línies anteriors que ens mostressin missatges d'error que ens indiquen l'origen d'aquest i el seu context. D'aquesta manera, podem realitzar un seguiment de pila. A més, trobarem el número de línia i el nom de l'arxiu o funció. Al mòdul següent (2d) s'aprofundirà en la lectura d'errors a Python.

### Tipus d'excepcions
 Hi ha excepcions de diferents tipus, i el tipus s'imprimeix com a part del missatge. Les principals excepcions definides a Python són:

*   `TypeError` : Ocorre quan s'aplica una operació o funció a una dada del tipus inapropiat.
*   `ZeroDivisionError` : Ocorre quan s'intenta dividir per zero.
*   `OverflowError` : Ocorre quan un càlcul excedeix el límit per a un tipus de dada numèrica.
*   `IndexError` : Ocorre quan s'intenta accedir a una seqüència amb un índex que no existeix.
*   `KeyError` : Ocorre quan s'intenta accedir a un diccionari amb una clau que no existeix.
*   `FileNotFoundError` : Ocorre quan s'intenta accedir a un fitxer que no existeix a la ruta indicada.
*   `ImportError` : Ocorre quan falla la importació d'un mòdul.

### Captura i llançament
Quan una eina escriu un missatge d'error, es genera una excepció.

Python permet escriure una rutina que s'executa automàticament quan es genera un error del sistema. En aquesta rutina de tractament dels errors, es captura el missatge d'error i s'actua en conseqüència; ja sigui mostrant un missatge personalitzat quan es produeixi un error o bé executant algun codi addicional.

Per al control d'excepcions a Python, es poden utilitzar les declaracions `try`, `except`, `else` i `finally` per manejar errors i excepcions de manera més complexa. Aquestes declaracions es poden utilitzar juntes o per separat, depenent de les necessitats del programa. A continuació es proporciona una breu descripció de cadascuna d'aquestes declaracions:

#### Try

`Try` permet al programa provar codi potencialment problemàtic. Si es produeix una excepció durant l'execució d'aquest codi, se saltarà al bloc `except` corresponent.

#### Except

`Except` permet al programa capturar qualsevol excepció que ocorri durant l'execució del codi al bloc `try`. Es pot especificar un tipus d'excepció específic per manejar, o es pot utilitzar una clàusula `except` genèrica per capturar qualsevol tipus d'excepció.

#### Else

`Else` s'executa si no es produeix cap excepció durant l'execució del codi al bloc `try`. Això és útil per executar codi addicional únicament si el codi al bloc `try` s'executa correctament.

#### Finally

`Finally` s'executa sempre, independentment de si es produeix una excepció o no durant l'execució del codi al bloc `try`. Això és útil per executar codi de neteja o de tancament que ha d'executar-se sempre, independentment del resultat del codi al bloc `try`.

A continuació, es pot observar la utilització de les declaracions:

```
try:
   # Codi potencialment problemàtic
except ExceptionType:
   # Codi per manejar l'excepció ExceptionType
else:
   # Codi a executar si no es produeix cap excepció
finally:
   # Codi a executar sempre al final de la declaració "try"
```

En aquest exemple, el codi al bloc `try` s'executarà primer. Si es produeix una excepció de tipus ExceptionType durant l'execució d'aquest codi, se saltarà al bloc `except` i s'executarà el codi per gestionar l'excepció. Si no es produeix cap excepció, se saltarà el bloc `except` i s'executarà el codi al bloc `else`. Finalment, el flux passarà pel codi dins de `finally`.

Exemples de control d'excepcions:

````python
>>> def division(num_1, num_2):
...     try:
...         result = num_1 / num_2
...     except ZeroDivisionError:
...         print('No es pot dividir per zero!')
...     else:
...         print(result)
...
>>> division(2, 0)
>>> division(15, 3)
````

````python
>>> try:
...     f = open('fitxer.txt')  # No existeix fitxer
... except FileNotFoundError:
...     print('No existeix fitxer!')
... else:
...     print(f.read())
````

**Capturar errors a Python identificant el tipus d'error**

Per afinar una mica més i mostrar missatges personalitzats a l'usuari o executar codi quan es produeix un error específic, d'algun tipus concret, es pot fer de la següent manera:

````python
try:
    numero_1 = int(input("Introdueix un nombre: "))
    numero_2 = int(input("Introdueix un segon nombre: "))
    division = numero_1 / numero_2
    print("{} / {} = {}".format(numero_1, numero_2, division))
except ValueError:
    print("Has d'introduir nombres")
except ZeroDivisionError:
    print("No es pot dividir un nombre entre zero")
````

A l'exemple anterior veiem que s'han capturat dos tipus d'excepcions:

*   **ValueError:** que serveix per capturar un possible error en la conversió de nombres. A l'exemple, si l'usuari no introdueix un nombre vàlid, es capturarà amb aquesta excepció i es mostrarà el missatge «Has d'introduir nombres».

*   **ZeroDivisionError:** es captura un possible error si l'usuari introdueix, en el segon nombre, un 0, atès que no és possible dividir un nombre entre zero, per la qual cosa el capturem i mostrem el missatge «No es pot dividir un nombre entre zero».


**Declaració raise**

En alguns casos, pot ser necessari crear excepcions personalitzades. Per a aquest propòsit, es pot utilitzar una declaració raise per llançar una excepció si ocorre una condició.

Per exemple:
```
raise Exception(‘Missatge’)
raise ValueError(‘Missatge’)
raise TypeError(‘Missatge’)
raise NameError(‘Missatge’)
```



````python
numero = 10
if numero > 5:
    raise Exception('El valor de número no ha d'excedir de 5. El valor va ser: {}'.format(numero))
````

### Errors comuns (gestió incorrecta de les excepcions)

**Accedir a posició no existent en array**

Aquest error fa referència a un índex d'un array fora de rang. Per exemple, si s'intenta accedir a l'element de l'array amb posició 100 però l'array només consta de tres elements, Python llançarà un `IndexError` indicant que l'índex de l'array està fora de rang.

Exemple:

````python
llista_amb_elements = [1, 2, True]
print(llista_amb_elements[100])
````



**Accedir a key no existent en un diccionari**

Quan no comprovem si una clau/key determinada es troba en un diccionari i intentem accedir-hi per mitjà d'aquesta clau, obtenim un dels errors més comuns i coneguts a Python: `KeyError`.

Ho veurem en un exemple:

````python
diccionari_de_prova = {"Gener": 125.96, "Febrer": 856.2}

print(diccionari_de_prova["Desembre"])
````

Aquest error pot ser controlat mitjançant la construcció `try`/`except` per actuar en conseqüència. Llavors s'ha d'escriure el codi que pot causar l'error dins del bloc `try` i el codi per gestionar l'error dins del bloc `except` de manera personalitzada.

Es pot afegir `else`, perquè el codi s'executi en cas que no hi hagi error. També, opcionalment es pot afegir `finally`, perquè el codi s'executi, hi hagi error o no.

Exemple:

````python
diccionari_de_prova = {"Gener": 125.96, "Febrer": 856.2}

clau = "Febrer"
try:
    print(f'El valor per a {clau} és {diccionari_de_prova[clau]}')  # s'executarà correctament
except:
    print(f'La clau {clau} no es troba al diccionari')

clau = "Desembre"
try:
    print(f'El valor per a {clau} és {diccionari_de_prova[clau]}')  # això genera un error KeyError
except:
    print(f'La clau {clau} no es troba al diccionari')
finally:
    print("Execució acabada i pas per finally")
````

## Proves bàsiques que cal tenir en compte

És important seguir certs procediments per comprovar que el codi funciona correctament. A continuació veurem alguns depenent del tipus de dada.

### Tipus booleà

És important saber en quins casos un booleà és vertader i fals. Per això és clau recordar que, per defecte, un objecte es considera vertader tret que la seva classe defineixi o bé un mètode `bool()` que retorna False o un mètode `len()` que retorna zero, quan s'invoqui des d'aquest objecte.

D'altra banda, es tenen objectes que s'avaluen com a falsos:

*   Constants definides per tenir valor fals: `None` i `False`.
*   Qualsevol col·lecció o seqüència buida: `''`, `()`, `[]`, `{}`, `set()`, `range(0)`

### Tipus numèric

#### Punt flotant

Els tipus de dades numèriques han de respectar certes precisions. A Python, els tipus de dades `float` i `double` no existeixen de manera nativa. En lloc seu, Python utilitza el tipus de dades `float` per representar nombres de punt flotant, que són una aproximació dels nombres reals a l'ordinador, la qual cosa significa que pot emmagatzemar nombres amb una precisió de fins a 15 dígits decimals.

A continuació, es mostra un cas on 2 valors de punt flotant no coincideixen, malgrat que a simple vista sí que semblen coincidir; la diferència és mínima a causa de la manera en què s'emmagatzemen els nombres de punt flotant a la memòria de l'ordinador.

Exemple:

````python
x = 0.1 + 0.1 + 0.1
y = 0.3

print(x == y)  # Imprimeix "False"
````

Per a aquest exemple en concret es pot solucionar amb llibreria com a Decimal.

#### Valors extrems

A Python, el tipus de dades `int` s'utilitza per representar enters (nombres enters, sense decimals). Aquests enters poden ser de qualsevol mida, fins al límit de memòria disponible a l'ordinador. Això significa que el tipus de dades `int` a Python té una precisió il·limitada, la qual cosa el fa adequat per a aplicacions que requereixin una precisió màxima per a nombres enters grans.

No obstant això, si es tracta d'utilitzar un enter molt gran, és possible que s'esgotin els recursos de memòria disponibles a l'ordinador i es produeixi un error de memòria, com en el següent exemple:

````python
x = 12345678987654321234567898765432123456789876543212345678987654322345678987654321234567898765432

print(x * x)  # Produeix un error de memòria
````

En aquest exemple, s'està tractant de multiplicar un enter extremadament gran per si mateix. Això requereix una quantitat significativa de memòria per emmagatzemar el resultat i, si no hi ha prou memòria disponible, es produirà un error de memòria.
