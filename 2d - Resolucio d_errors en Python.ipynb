### Exemples d'errors de Python

Els errors més comuns són:

1. `SyntaxError`: Aquest error ocorre quan es troba un error de sintaxi al codi, una estructura malament escrita o un parèntesi que no es trobi tancat.
2. `NameError`: Es produeix en el moment d'ocupar una variable que no ha estat definida prèviament.
3. `TypeError`: Ocorre quan els tipus de dades no són els correctes.
4. `IndexError`: Ocorre quan s'ocupa un índex fora del rang existent.
5. `ZeroDivisionError`: Ocorre quan es divideix un nombre per 0, que matemàticament és considerada una indefinició.
