# Gramàtica i sintaxi de Python

La gramàtica de Python és molt clara i senzilla de seguir, i es basa en un conjunt de regles ben definides. De fet, la guia d'estil de Python és ben coneguda i està disponible en el document [PEP8](https://peps.python.org/pep-0008/).

## Operadors

Python té un conjunt d'operadors aritmètics i lògics que es poden utilitzar per fer càlculs i comparacions al codi.

### Operadors aritmètics
Python té un conjunt d'operadors aritmètics que es poden utilitzar per realitzar càlculs matemàtics. Alguns exemples d'operadors aritmètics són:

````python
a = 10
b = 5

c = a + b  # c conté el resultat de a + b (15)
d = a - b  # d conté el resultat de - b (5)
e = a * b  # e conté el resultat de * b (50)
f = a / b  # f conté el resultat de / b (2.0)
````

### Operadors de comparació o lògics

Python té un conjunt d'operadors de comparació que es poden utilitzar per comparar dos valors i obtenir un resultat booleà (vertader o fals). Alguns exemples d'operadors de comparació són:

````python
x = 10
y = 20

resultat = x < y  # el resultat de x < y (vertader)
resultat = x > y  # el resultat de x > y (fals)
resultat = x == y  # el resultat de x == y (fals)
resultat = x != y  # el resultat de x != y (vertader)
````

### Múltiples línies

PEP8 recomana que les línies no excedeixin els 79 caràcters. Fent ús de \ es pot distribuir el codi en diverses línies, la qual cosa en determinats casos fa que el codi sigui molt més llegible.

```

    5 + 3 + 6 + 9
```

Si pel contrari estem dins d'un bloc envoltat amb parèntesis (), n'hi hauria prou amb saltar a la següent línia.

```

     5 + 3 + 6 + 9)
```     

## **1. Regles de tipatge a Python**
Python és un llenguatge de tipatge **dinàmic** i **fort**.

És de **tipatge dinàmic**, ja que una variable pot començar tenint un tipus de dada i pot canviar en qualsevol moment a un altre tipus de dada. Per exemple:

````python
variable_dinamica = 10
print(variable_dinamica)
variable_dinamica = "Hola món"
print (variable_dinamica)
````

Com s'ha pogut observar anteriorment, la variable_dinamica és creada amb el valor de 10, que és un nombre enter (int). Després, a la tercera línia s'assigna el nou valor "Hola món", per la qual cosa el tipus de dada canvia a una cadena (str).

D'altra banda, Python és un llenguatge de **tipatge fort**; ja que davant una operació entre dos tipus de dades incompatibles, llança un error (durant la compilació o l'execució) en lloc de convertir implícitament algun dels dos tipus. Per exemple:

````python
a = 10
b = "5"
print (a + b)
````

## **2. Aprofundiment en les estructures bàsiques**

Python compta amb bucles iteratius com `for` i `while`; no obstant això, en altres llenguatges de programació n'hi ha d'altres com `do while`.

Llavors, quines són les seves diferències? Quan hauria d'usar un en lloc d'un altre?

### FOR
S'utilitza quan ja es coneix el nombre d'iteracions, és a dir, quan se sap quantes vegades s'ha d'executar una instrucció . És per això que s'usa amb objectes finits com ho és una llista, un diccionari, un set, etc.



### WHILE
En canvi, el bucle `while` es fa servir quan el nombre d'iteracions pot variar depenent de dos factors:


*   Dels valors que es “comparen” de manera inicial, podent ser que a causa de la condició no s'arribin a executar.
*   Les operacions que es realitzin dins del mateix cicle.

Amb `while`, mentre la condició sigui certa s'executa el cos del bucle. Si la condició és falsa acaba el bucle. Només executa el seu contingut si es comprova una condició, la qual cosa pot ocórrer 0, 1 o més vegades.

### DO WHILE
Específicament no es compta amb el bucle `do while` en Python, però es pot construir unint un bucle `while` i una condició.

Igual que `while`, el bucle `do while` s'utilitza quan el nombre d'iteracions varia; no obstant això, la instrucció `do while` permet repetir una instrucció fins que una expressió especificada sigui falsa. El codi s'executa almenys una vegada encara que la condició sigui falsa, perquè la condició és verificada després de l'execució del codi.



### Exemple de conversió de while / do while / for per al factorial d'un número:

#### FOR

````python
numero = 4
factorial = 1
for i in range (1,numero + 1):
    factorial = factorial * i
print ("El factorial del nombre",numero,"és:",factorial)
````

#### WHILE

````python
numero = 4
i = 0
factorial = 1
while i < numero:
    factorial = factorial * (i+1)
    i = i + 1
print ("El factorial del nombre", numero,"és:",factorial)
````

#### DO WHILE

````python
numero = 4
i = 0
factorial = 1
while True:
    factorial = factorial * (i+1)
    i = i + 1
    if numero <= i:
        break
print ("El factorial del nombre", numero,"és:",factorial)
````

## **3. Àmbit de les variables**
L'àmbit d'una variable és el context en el qual existeix aquesta variable. Per això, una variable existeix en aquest àmbit a partir del moment en què es crea i deixa d'existir quan desapareix el seu àmbit. Els principals tipus d'àmbits a Python són dos: àmbit local i àmbit global.

Per entendre millor els tipus d'àmbits ho farem mitjançant funcions; recordem que les funcions a Python constitueixen unitats lògiques d'un programa i tenen un doble objectiu:


### Variables globals
Les variables globals són totes les que es creen fora d'una funció. Aquestes variables poden ser utilitzades en tot el codi, tant dins com fora de les funcions.

Exemple:

````python
variable_global = "global"

def funcion():
    print("Aquesta és una variable "+ variable_global)

funcion()
````

### Funció
L'àmbit d'una variable dins d'una funció només existeix dins de la funció; aquestes variables són conegudes com a variables locals o d'àmbit local. Cal tenir en compte que si la variable local té el mateix nom d'una variable global, el valor d'aquesta última no es modificarà.

Exemple:

````python
variable = "global"

def funcion():
    variable = "local"
    print("Aquesta és una variable",variable)

funcion()
print("Aquesta és una variable",variable)
````

Es poden importar/invocar variables globals des d'altres mòduls o arxius fent un ús adequat de l'import.

### Paquet
L'àmbit de les variables d'un paquet existirà dins del paquet i dins del codi en el qual es realitzi la crida/importació del paquet.

## **4. Pas per valor i pas per referència**

Depenent del tipus de dada que s'enviï a la funció, es poden diferenciar dos comportaments:
* Pas per valor: es crea una còpia local de la variable dins de la funció.
* Pas per referència: es maneja directament la variable, i els canvis realitzats dins de la funció l'afectaran també fora. A continuació s'explicarà amb deteniment:

### **Pas per valor**
Quan s'utilitza pas per valor dels arguments, el que es fa és copiar el valor de les variables als respectius paràmetres. Així doncs, qualsevol modificació del valor del paràmetre no afecta la variable externa corresponent. Generalment, els tipus simples es passen per valor: enters, flotants, cadenes, lògics, etc.

Exemple:

````python
def doble_numero(numero):
    numero = numero * 2

num = 10
doble_numero(num)
print(num)
````

### **Pas per referència**
Quan s'utilitza pas per referència, el que es fa és copiar en els paràmetres l'adreça de memòria de les variables que s'usen com a argument. Això implica que realment facin referència al mateix objecte/element i qualsevol modificació del valor en el paràmetre afectarà la variable externa corresponent. Generalment els tipus composts es passen per referència: llistes, diccionaris, conjunts, etc.

Exemple:

````python
def doble_numero(numeros):
    for i,n in enumerate(numeros):
        numeros[i] *= 2

n = [10,50,100]
doble_numero(n)
print(n)
````
